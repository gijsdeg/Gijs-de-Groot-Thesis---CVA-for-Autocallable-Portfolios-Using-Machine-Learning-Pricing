```python
# Exposure plots
import pickle, glob, os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from numpy.random import default_rng

from scipy.interpolate import CubicSpline


path = r"C:\Users\14gij\OneDrive\Documenten\1 Master\Thesis\thesis code\CVA\exposure_and_cva_files"



with open(os.path.join(path, f"exposures_final_en_2025-07-26.pkl"), "rb") as fh:
        expos_en = pickle.load(fh)


with open(os.path.join(path, f"exposures_final_xgb_2025-07-26.pkl"), "rb") as fh:
        expos_xgb = pickle.load(fh)

        
with open(os.path.join(path, f"exposures_final_nn_2025-07-26.pkl"), "rb") as fh:
        expos_nn = pickle.load(fh)

        
with open(os.path.join(path, f"exposures_final_gpr_2025-07-26.pkl"), "rb") as fh:
        expos_gpr = pickle.load(fh)


with open(os.path.join(path, "pd_final_2025-07-26.pkl"), "rb") as fh:
        pd_paths = pickle.load(fh)


with open(os.path.join(path, f"cva_final_2025-07-26.pkl"), "rb") as fh:
        cva = pickle.load(fh)


NUM_STEPS = 52 * 5 * 7 + 2
PRICE_POINTS = []
workdays_per_quarter = (365*4+1)/7*5/4/4  # +1 leap day every four years
i = 1
while True:
    t = round(workdays_per_quarter * i)
    if t >= NUM_STEPS:
        break
    PRICE_POINTS.append(t)
    i += 1
PRICE_POINTS = np.array(PRICE_POINTS)
NUM_PRICES = len(PRICE_POINTS)


# def discount_rates(price_points):
#     # Connect to the database
#     conn = DBConnection(engine=get_database_connection_sp())
    
#     # Query the most recent yield curve
#     curve_query = '''SELECT TOP 1
#             [1BD], [1WK], [1MO], [2MO], [3MO], [4MO], [5MO], [6MO], [7MO], [8MO],
#             [9MO], [10MO], [11MO], [12MO], [18MO], [2YR], [3YR], [4YR], [5YR], [7YR],
#             [10YR], [12YR], [15YR], [20YR], [25YR], [30YR], [40YR], [50YR]
#         FROM StructuredProducts2010.dbo.YieldCurve
#         WHERE RateId = 56 AND CAST(ModifyDate as date) = '2025-07-23'
#         ORDER BY ModifyDate DESC'''
    
#     # Fetch the yield curve data
#     recent_curve = conn.fetch(curve_query)
    
#     # Define the maturities corresponding to the yield curve data (in years)
#     maturities = np.array([1/365, 7/365, 1/12, 2/12, 3/12, 4/12, 5/12, 6/12, 
#                            7/12, 8/12, 9/12, 10/12, 11/12, 1, 1.5, 2, 3, 4, 
#                            5, 7, 10, 12, 15, 20, 25, 30, 40, 50])  
    
#     # Extract the yield curve rates (assuming the data is in the first row)
#     yield_rates = recent_curve.iloc[0].values / 100  # Convert from percentage to decimals
    
#     # Apply cubic spline interpolation
#     days = maturities * 365  # Convert maturities to days for interpolation
#     spline = CubicSpline(days, yield_rates, extrapolate=True)
    
#     # Interpolated yield rates for the given price_points (in days)
#     interpolated_yield_rates = spline(price_points)
    
#     # Convert interpolated yield rates to discount factors using continuous compounding
#     discount_factors = np.exp(-interpolated_yield_rates * (price_points / 365))
    
#     return discount_factors

# disc = discount_rates(PRICE_POINTS)
    

with open(os.path.join(path, f"discount_jonge.pkl"), "rb") as fh:
    disc = pickle.load(fh)


# 27 rows × 3 columns, no special column names or index
data = [
    [0.0019113852578894752, 0.002546395278435032,  0.0029668995525442554],
    [0.0018788540594237846, 0.0025014759436118306, 0.002913342953322684 ],
    [0.001904140650334396,  0.0025335414029890035, 0.0029494534636846925],
    [0.0018717327528963157, 0.0024888488151835197, 0.002896211790254233 ],
    [0.0018682093055502058, 0.002482607125937064,  0.0028877490238661263],
    [0.001864692490924913,  0.002476381089985491,  0.0028793109857850885],
    [0.0018897884881129423, 0.00250812486807217,   0.0029149996742963147],
    [0.0018576248601310796, 0.0024638806371466426, 0.002862379938938897 ],
    [0.0018541279702178892, 0.002457701564643755,  0.002854016029635309 ],
    [0.0018506376630327281, 0.0024515379884018085, 0.0028456765597772993],
    [0.001847153926183731,  0.0024453898695581122, 0.0028373614579524364],
    [0.0018720138802850173, 0.002476736383092981,  0.0028725301874735765],
    [0.0018401527707475607, 0.0024330458564087998, 0.002820677084502843 ],
    [0.0018366887712066848, 0.0024269441132792613, 0.0028124350314514546],
    [0.0018332312924790317, 0.0024208576724793485, 0.0028042170617804674],
    [0.001857903868566102,  0.002451889717202005,  0.0028389749707836698],
    [0.0018262829071565845, 0.0024086374947016598, 0.0027877275854180095],
    [0.0018228450169701649, 0.0024025969643739487, 0.00277958181120741  ],
    [0.0018194135984476167, 0.002396571582862439,  0.0027714598390490286],
    [0.0018159886394062408, 0.002390561312175743,  0.002763361599392833 ],
    [0.0018404291549387608, 0.0024212049994916107, 0.0027976131102063206],
    [0.0018091056078652912, 0.002378494066523573,  0.002747112363057136 ],
    [0.0018057000531229717, 0.0023725291317523434, 0.002739085267024488 ],
    [0.0018023009091743258, 0.002366579156214077,  0.0027310816262651194],
    [0.0018265572081452852, 0.0023969154254835923, 0.002764933030972694 ],
    [0.0017954697574067424, 0.0023546329693967527, 0.002715022203999329 ],
    [0.0017920898715004796, 0.002348727874963208,  0.002707088875805286 ]
]

static_pd = pd.DataFrame(data)


# --------------------------------------------------------------
# 🔺 FILL ME 🔺  ---  load your objects here --------------------
# --------------------------------------------------------------
exposures_models = [expos_en, expos_xgb, expos_nn, expos_gpr]   # [expos_en, expos_xgb, expos_nn, expos_gpr]  (len=4)
pd_paths_dyn     = pd_paths   # list (n_sims) of DataFrames 27x3  (dynamic conditional PDs)
discount         = disc     # length‑27 NumPy array
static_pd_df     = static_pd     # 27x3 DataFrame  (constant conditional PD paths)
cva_dict         = cva     # {'en': df1000x3, 'xgb': ... , 'nn': ..., 'gpr': ...}
# --------------------------------------------------------------

model_names = ["en", "xgb", "nn", "gpr"]
cp_names    = static_pd_df.columns.tolist()
lgd         = 0.6
alpha       = 0.95   # VaR/ES level
rng         = default_rng(42)


# ---------- 0. Tell the script the PDs are already unconditional ------------
PD_ARE_UNCOND = True          # <<<<< set to False if you later pass conditional q_i

# ---------- helpers ---------------------------------------------------------
def cond_to_uncond(q):
    """Convert conditional bucket PDs to unconditional ΔPD.
       If PD_ARE_UNCOND is True, just return q unchanged."""
    if PD_ARE_UNCOND:
        return q               # already ΔPD_i
    # else convert q_i -> S_{i-1} q_i
    d = np.empty_like(q)
    surv = 1.0
    for i, qi in enumerate(q):
        d[i] = surv * qi
        surv *= (1 - qi)
    return d


def stack(mat_list, cp_idx, pd=False):
    if pd:
        return np.stack([df.iloc[:, cp_idx].to_numpy() for df in mat_list], axis=1)
    return np.stack([exp[:, cp_idx] for exp in mat_list], axis=1)

def cva_vec(exp_list, pd_list, cp_idx):
    """joint CVA per simulation"""
    n = len(exp_list)
    cva = np.empty(n)
    for i, (exp, df) in enumerate(zip(exp_list, pd_list)):
        dPD = cond_to_uncond(df.iloc[:, cp_idx].to_numpy())
        cva[i] = lgd * np.sum(discount * exp[:, cp_idx] * dPD)
    return cva

def static_cva_vec(exp_list, static_pd_col):
    dPD_static = cond_to_uncond(static_pd_col.to_numpy())
    return np.array([lgd * np.sum(discount * exp[:,cp]*dPD_static)
                     for exp in exp_list])

def indep_cva(exp_list, pd_list, cp_idx):
    """product‑of‑means independence CVA (scalar)"""
    exp_mat = stack(exp_list, cp_idx)             # (27,N)
    q_mat   = stack(pd_list, cp_idx, pd=True)     # (27,N)
    dPD_mat = np.column_stack([cond_to_uncond(q_mat[:,j]) for j in range(q_mat.shape[1])])
    mean_exp = exp_mat.mean(axis=1)
    mean_dpd = dPD_mat.mean(axis=1)
    return lgd * np.sum(discount * mean_exp * mean_dpd)

def var_es(x, alpha=0.95):
    x = np.sort(x)
    idx = int(np.floor(alpha*len(x))) - 1
    var = x[idx]
    es  = x[idx:].mean()
    return var, es



# ---------- main loops -------------------------------------------------------
rows_chk, rows_dyn, rows_stat, rows_indep = [], [], [], []
rows_uplift, rows_level, rows_wwr = [], [], []

z95 = 1.96                                # two‑sided 95 % normal quantile

for m_idx, m in enumerate(model_names):
    exp_list = exposures_models[m_idx]
    dyn_df   = cva_dict[m]
    for cp_idx, cp in enumerate(cp_names):

        # 1) recompute joint CVA .............................
        cva_joint_vec = cva_vec(exp_list, pd_paths_dyn, cp_idx)
        rows_chk.append([cp, m, cva_joint_vec.mean()])

        # 2) dynamic CVA stats + 95 % CI .....................
        dyn_vals = dyn_df.iloc[:, cp_idx].to_numpy()
        mean_dyn = dyn_vals.mean()
        sd_dyn   = dyn_vals.std(ddof=1)
        se_dyn   = sd_dyn / np.sqrt(len(dyn_vals))
        ci_lo_dyn, ci_hi_dyn = mean_dyn - z95*se_dyn, mean_dyn + z95*se_dyn
        v95_dyn, e95_dyn = var_es(dyn_vals, alpha)
        rows_dyn.append([cp, m, mean_dyn, sd_dyn, ci_lo_dyn, ci_hi_dyn, v95_dyn, e95_dyn])

        # 3) static CVA stats + 95 % CI .....................
        stat_vec = static_cva_vec(exp_list, static_pd_df.iloc[:, cp_idx])
        mean_stat = stat_vec.mean()
        sd_stat   = stat_vec.std(ddof=1)
        se_stat   = sd_stat / np.sqrt(len(stat_vec))
        ci_lo_stat, ci_hi_stat = mean_stat - z95*se_stat, mean_stat + z95*se_stat
        v95_stat, e95_stat = var_es(stat_vec, alpha)
        rows_stat.append([cp, m, mean_stat, sd_stat, ci_lo_stat, ci_hi_stat, v95_stat, e95_stat])

        # 4) independence CVA ................................
        cva_indep = indep_cva(exp_list, pd_paths_dyn, cp_idx)
        rows_indep.append([cp, m, cva_indep])

        # 5) decomposition ...................................
        d_total = mean_dyn  - mean_stat
        d_level = cva_indep - mean_stat
        d_wwr   = mean_dyn  - cva_indep
        rows_uplift.append([cp, m, d_total])
        rows_level.append ([cp, m, d_level])
        rows_wwr.append   ([cp, m, d_wwr])

# ---------- build tables -----------------------------------------------------
def mk(df, cols):
    return pd.DataFrame(df, columns=["CP","Model"]+cols).set_index(["CP","Model"])

T1 = mk(rows_chk,  ["CVA_joint_recalc"])

T2 = mk(
    rows_dyn,
    ["Mean","SD","CI95_low","CI95_high","VaR95","ES95"]
)

T3 = mk(
    rows_stat,
    ["Mean","SD","CI95_low","CI95_high","VaR95","ES95"]
)

T4 = mk(rows_indep,["CVA_indep"])
T5 = mk(rows_uplift,["Δ_total"])
T6 = mk(rows_level, ["Δ_level"])
T7 = mk(rows_wwr,   ["Δ_WWR"])

# ---------- print ------------------------------------------------------------
pd.set_option("display.float_format", "{:,.6f}".format)

print("\n=== T1. Re‑computed mean CVA (dyn‑joint) ===")
print(T1)

print("\n=== T2. Dynamic CVA distribution stats ===")
print(T2)

print("\n=== T3. Static‑PD CVA distribution stats ===")
print(T3)

print("\n=== T4. Independence CVA (product‑of‑means) ===")
print(T4)

print("\n=== T5. Δ_total = dyn_joint − static ===")
print(T5)

print("\n=== T6. Δ_level/timing = dyn_indep − static ===")
print(T6)

print("\n=== T7. Pure WWR = dyn_joint − dyn_indep ===")
print(T7)
```

```python
# Exposure plots
import pickle, glob, os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from numpy.random import default_rng



path = r"C:\Users\UGGROO\OneDrive - Van Lanschot Kempen\Documents\MSc Thesis\3. Repos\thesis code\CVA\exposure_and_cva_files"


with open(os.path.join(path, f"exposures_retuned_en_2025-07-23.pkl"), "rb") as fh:
        exposures_en = pickle.load(fh)


with open(os.path.join(path, f"exposures_retuned_xgb_2025-07-23.pkl"), "rb") as fh:
        exposures_xgb = pickle.load(fh)

        
with open(os.path.join(path, f"exposures_retuned_nn_2025-07-23.pkl"), "rb") as fh:
        exposures_nn = pickle.load(fh)

        
with open(os.path.join(path, f"exposures_retuned_gpr_2025-07-23.pkl"), "rb") as fh:
        exposures_gpr = pickle.load(fh)


with open(os.path.join(path, "pd_retuned_2025-07-23.pkl"), "rb") as fh:
    pd_paths = pickle.load(fh)


# ---------- utilities ----------
def fisher_pool(r, n_eff=None, eps=1e-12):
    r = np.asarray(r, float)
    r = np.clip(r, -1 + eps, 1 - eps)
    z = np.arctanh(r)
    if n_eff is None:
        z_bar = np.nanmean(z)
    else:
        w = np.asarray(n_eff) - 3  # classic weight for correlations
        w = np.where(np.isnan(z), 0.0, w)
        z_bar = np.nansum(w * z) / np.sum(w)
    return float(np.tanh(z_bar))

def bootstrap_ci(values, n_boot=2000, alpha=0.05, random_state=42, pool_with=None, n_eff=None):
    rng = default_rng(random_state)
    values = np.asarray(values)
    idx = ~np.isnan(values)
    vals = values[idx]
    if n_eff is not None:
        n_eff = np.asarray(n_eff)[idx]
    boot = []
    n = len(vals)
    for _ in range(n_boot):
        bidx = rng.integers(0, n, n)
        if pool_with == "fisher":
            boot.append(fisher_pool(vals[bidx], None if n_eff is None else n_eff[bidx]))
        else:
            boot.append(np.nanmean(vals[bidx]))
    lo, hi = np.quantile(boot, [alpha/2, 1 - alpha/2])
    return lo, hi

# ---------- per-simulation correlations ----------
def per_sim_corrs(pd_paths, exposures_model, cp_idx,
                  method="pearson", trim_trailing_zeros=True):
    rs, n_eff = [], []
    for pd_df, exp in zip(pd_paths, exposures_model):
        p = pd_df.iloc[:, cp_idx].to_numpy()
        x = exp[:, cp_idx]

        if trim_trailing_zeros:
            nz = np.flatnonzero(x != 0.0)
            if nz.size == 0:
                rs.append(np.nan); n_eff.append(0)
                continue
            last = nz[-1] + 1
            p = p[:last]; x = x[:last]

        if np.std(p) == 0 or np.std(x) == 0:
            rs.append(np.nan); n_eff.append(len(p))
            continue

        if method == "pearson":
            r = np.corrcoef(p, x)[0, 1]
        elif method == "spearman":
            r = pd.Series(p).corr(pd.Series(x), method="spearman")
        else:
            raise ValueError("method must be 'pearson' or 'spearman'")
        rs.append(r); n_eff.append(len(p))
    return np.array(rs, float), np.array(n_eff, int)

# ---------- cross-simulation, time-slice correlations ----------
def time_slice_corrs(pd_paths, exposures_model, cp_idx):
    # stack to (T, Nsim)
    pd_mat  = np.stack([df.iloc[:, cp_idx].to_numpy() for df in pd_paths], axis=1)
    exp_mat = np.stack([exp[:, cp_idx] for exp in exposures_model], axis=1)
    T = pd_mat.shape[0]
    out = np.full(T, np.nan)
    for t in range(T):
        x, y = pd_mat[t], exp_mat[t]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() > 1 and np.std(x[mask]) > 0 and np.std(y[mask]) > 0:
            out[t] = np.corrcoef(x[mask], y[mask])[0, 1]
    return out

# ---------- aggregate-path correlation ----------
def aggregate_path_corr(pd_paths, exposures_model, cp_idx, weight=None):
    # Reduce each path to a scalar and correlate across simulations
    P = np.array([df.iloc[:, cp_idx].to_numpy() for df in pd_paths])  # (N, T)
    X = np.array([exp[:, cp_idx] for exp in exposures_model])         # (N, T)
    if weight is None:
        p_scalar = P.mean(axis=1)
        x_scalar = X.mean(axis=1)
    else:
        w = np.asarray(weight) / np.sum(weight)
        p_scalar = P @ w
        x_scalar = X @ w
    if np.std(p_scalar) == 0 or np.std(x_scalar) == 0:
        return np.nan
    return float(np.corrcoef(p_scalar, x_scalar)[0, 1])

# ---------- example driver producing the 12 summary numbers ----------
def summarize_dep(pd_paths, exposures_models, model_names=("EN","XGB","NN","GPR"),
                  method="pearson", trim_trailing_zeros=True, n_boot=2000):
    n_cp = pd_paths[0].shape[1]
    table = pd.DataFrame(index=[f"CP{i+1}" for i in range(n_cp)],
                         columns=model_names, dtype=float)
    ci_lo = table.copy(); ci_hi = table.copy()
    med = table.copy(); iqr = table.copy()

    for cp_idx in range(n_cp):
        for name, exposures in zip(model_names, exposures_models):
            r_vec, n_eff = per_sim_corrs(pd_paths, exposures, cp_idx,
                                         method=method,
                                         trim_trailing_zeros=trim_trailing_zeros)

            pooled = fisher_pool(r_vec, n_eff)
            lo, hi = bootstrap_ci(r_vec, n_boot=n_boot, pool_with="fisher", n_eff=n_eff)
            table.loc[f"CP{cp_idx+1}", name] = pooled
            ci_lo.loc[f"CP{cp_idx+1}", name] = lo
            ci_hi.loc[f"CP{cp_idx+1}", name] = hi
            med.loc[f"CP{cp_idx+1}", name] = np.nanmedian(r_vec)
            iqr.loc[f"CP{cp_idx+1}", name] = np.nanpercentile(r_vec, 75) - np.nanpercentile(r_vec, 25)

    return table, (ci_lo, ci_hi), med, iqr

# ---- call it (assuming you already loaded pd_paths & the 4 exposure lists) ----
avg, (lo, hi), med, iqr = summarize_dep(
    pd_paths,
    [exposures_en, exposures_xgb, exposures_nn, exposures_gpr],
    model_names=["EN","XGB","NN","GPR"],
    method="pearson",
    trim_trailing_zeros=True,
    n_boot=2000
)

print("Fisher-pooled mean correlations:")
print(avg.round(3))
print("\n95% bootstrap CI:")
print(lo.round(3))
print(hi.round(3))
print("\nMedian and IQR of per-simulation correlations:")
print(med.round(3))
print(iqr.round(3))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy.random import default_rng
from marketdata import discount_rates

# ===============================================================
# 0) CONFIG
# ===============================================================
RECOVERY = 0.40                 # change if needed
LGD = 1.0 - RECOVERY
N_BOOT = 2000
ALPHA = 0.05
TAIL_Q = 0.90                   # top 10% for tail WWR
METHOD_TIME_SLICE = "spearman"  # or "pearson"
METHOD_PATHWISE = "pearson"     # for per-simulation path correlations
TRIM_TRAILING_ZEROS = True

# Provide a discount factor vector of length T (per time step).
# If you don't have one yet, set to np.ones(T)
def flat_discount(T, dt=0.25, r=0.0):
    # discount paid at the END of each step
    times = (np.arange(T) + 1) * dt
    return np.exp(-r * times)

# ===============================================================
# 1) HELPERS
# ===============================================================
def condpd_to_survival_and_dpd(q):
    """
    q: conditional default probability per step (length T)
    Returns:
        S: survival prob *after* the step (length T)
        dPD: unconditional default probability *during* the step (length T)
    """
    q = np.clip(q, 0.0, 1.0)
    T = len(q)
    S = np.empty(T)
    dPD = np.empty(T)
    S_prev = 1.0
    for t in range(T):
        dPD[t] = S_prev * q[t]
        S[t] = S_prev * (1.0 - q[t])
        S_prev = S[t]
    return S, dPD

def fisher_pool(r, n_eff=None, eps=1e-12):
    r = np.asarray(r, float)
    r = r[~np.isnan(r)]
    if r.size == 0:
        return np.nan
    r = np.clip(r, -1 + eps, 1 - eps)
    z = np.arctanh(r)
    if n_eff is None:
        z_bar = np.mean(z)
    else:
        w = np.asarray(n_eff) - 3
        w = np.maximum(w, 0.0)
        z_bar = np.sum(w * z) / np.sum(w)
    return float(np.tanh(z_bar))

def bootstrap_ci(vec, n_boot=2000, alpha=0.05, random_state=42, func=np.mean):
    rng = default_rng(random_state)
    vec = np.asarray(vec)
    vec = vec[~np.isnan(vec)]
    if vec.size == 0:
        return np.nan, np.nan
    n = len(vec)
    boots = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, n)
        boots.append(func(vec[idx]))
    lo, hi = np.quantile(boots, [alpha/2, 1 - alpha/2])
    return lo, hi

def make_const_pd_paths_from_scalar(pd_paths, const_pd_per_cp):
    """
    Build a 'constant PD' dataset if you don't already have one.
    const_pd_per_cp: iterable of length n_cp with the constant PD per step (quarter).
    """
    n_sim = len(pd_paths)
    T, n_cp = pd_paths[0].shape
    out = []
    for _ in range(n_sim):
        df = pd.DataFrame(
            np.tile(const_pd_per_cp, (T, 1)),
            columns=[f"CP{i+1}" for i in range(n_cp)]
        )
        out.append(df)
    return out

def stack_paths_pd(pd_paths, cp_idx):
    return np.stack([df.iloc[:, cp_idx].to_numpy() for df in pd_paths], axis=1)  # (T, N)

def stack_paths_exp(exposures_model, cp_idx):
    return np.stack([exp[:, cp_idx] for exp in exposures_model], axis=1)  # (T, N)

# ===============================================================
# 2) CORE METRICS
# ===============================================================
def cva_per_sim(exposures_model, pd_paths_like, cp_idx, discount):
    """
    Returns per-simulation CVA (length N_sims) given exposure list and PD paths list.
    """
    n_sim = len(exposures_model)
    out = np.zeros(n_sim)
    for i, (exp, df) in enumerate(zip(exposures_model, pd_paths_like)):
        q = df.iloc[:, cp_idx].to_numpy()
        _, dPD = condpd_to_survival_and_dpd(q)
        out[i] = LGD * np.sum(discount * exp[:, cp_idx] * dPD)
    return out

def wwr_uplift_tables(pd_paths, pd_paths_const, exposures_models, model_names, discount):
    """
    Returns three DataFrames (mean uplift, lower CI, upper CI) of shape (3 counterparties x 4 models).
    """
    n_cp = pd_paths[0].shape[1]
    uplift_mean = pd.DataFrame(index=[f"CP{i+1}" for i in range(n_cp)], columns=model_names, dtype=float)
    uplift_lo   = uplift_mean.copy()
    uplift_hi   = uplift_mean.copy()

    for cp_idx in range(n_cp):
        for name, exposures in zip(model_names, exposures_models):
            cva_joint = cva_per_sim(exposures, pd_paths, cp_idx, discount)
            cva_base  = cva_per_sim(exposures, pd_paths_const, cp_idx, discount)
            ratio = cva_joint / cva_base - 1.0

            uplift_mean.loc[f"CP{cp_idx+1}", name] = np.nanmean(ratio)
            lo, hi = bootstrap_ci(ratio, n_boot=N_BOOT, alpha=ALPHA, func=np.mean)
            uplift_lo.loc[f"CP{cp_idx+1}", name] = lo
            uplift_hi.loc[f"CP{cp_idx+1}", name] = hi

    return uplift_mean, uplift_lo, uplift_hi

def cva_covariance_breakdown(pd_paths, exposures_model, cp_idx, discount):
    """
    Returns a DataFrame with columns: base, cov, total, share_cov, per time step.
    base_t = LGD * D_t * E[EXP_t] * E[dPD_t]
    cov_t  = LGD * D_t * Cov(EXP_t, dPD_t)
    total  = base + cov
    """
    # stack
    exp_mat = stack_paths_exp(exposures_model, cp_idx)  # (T, N)
    # compute dPD per sim first
    q_mat = stack_paths_pd(pd_paths, cp_idx)            # (T, N)
    T, N = q_mat.shape
    dPD_mat = np.zeros_like(q_mat)
    for j in range(N):
        _, dPD = condpd_to_survival_and_dpd(q_mat[:, j])
        dPD_mat[:, j] = dPD

    E_exp = np.nanmean(exp_mat, axis=1)
    E_dpd = np.nanmean(dPD_mat, axis=1)
    E_prod = np.nanmean(exp_mat * dPD_mat, axis=1)
    cov = E_prod - E_exp * E_dpd

    base = LGD * discount * E_exp * E_dpd
    cov_term = LGD * discount * cov
    total = base + cov_term
    share_cov = np.divide(cov_term, total, out=np.zeros_like(total), where=total!=0)

    return pd.DataFrame({
        "base": base,
        "cov": cov_term,
        "total": total,
        "share_cov": share_cov
    })

def time_slice_corrs(pd_paths, exposures_model, cp_idx, method="spearman", use_dpd=False):
    """
    Correlation across simulations for each time t.
    If use_dpd=True, correlates EXP_t with dPD_t; otherwise with PD_t.
    """
    exp_mat = stack_paths_exp(exposures_model, cp_idx)      # (T, N)
    pd_mat  = stack_paths_pd(pd_paths, cp_idx)              # (T, N)
    if use_dpd:
        T, N = pd_mat.shape
        dpd_mat = np.zeros_like(pd_mat)
        for j in range(N):
            _, dPD = condpd_to_survival_and_dpd(pd_mat[:, j])
            dpd_mat[:, j] = dPD
        y_mat = dpd_mat
    else:
        y_mat = pd_mat

    T = exp_mat.shape[0]
    out = np.full(T, np.nan)
    for t in range(T):
        x = exp_mat[t, :]
        y = y_mat[t, :]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() > 1 and np.std(x[mask]) > 0 and np.std(y[mask]) > 0:
            if method == "pearson":
                out[t] = np.corrcoef(x[mask], y[mask])[0, 1]
            else:
                out[t] = pd.Series(x[mask]).corr(pd.Series(y[mask]), method="spearman")
    return out

def per_sim_path_corrs(pd_paths, exposures_model, cp_idx,
                       method="pearson", trim_trailing_zeros=True):
    rs, n_eff = [], []
    for df, exp in zip(pd_paths, exposures_model):
        p = df.iloc[:, cp_idx].to_numpy()
        x = exp[:, cp_idx]
        if trim_trailing_zeros:
            nz = np.flatnonzero(x != 0.0)
            if nz.size == 0:
                rs.append(np.nan); n_eff.append(0)
                continue
            last = nz[-1] + 1
            p = p[:last]; x = x[:last]

        if np.std(p) == 0 or np.std(x) == 0:
            rs.append(np.nan); n_eff.append(len(p))
            continue

        if method == "pearson":
            r = np.corrcoef(p, x)[0, 1]
        else:
            r = pd.Series(p).corr(pd.Series(x), method="spearman")
        rs.append(r); n_eff.append(len(p))
    return np.array(rs), np.array(n_eff)

def tail_wwr_uplift(pd_paths, exposures_model, cp_idx, q=0.9, use_dpd=False):
    """
    For each time t: E[EXP_t | PD_t (or dPD_t) in top q] / E[EXP_t].
    Returns a vector length T.
    """
    exp_mat = stack_paths_exp(exposures_model, cp_idx)
    pd_mat  = stack_paths_pd(pd_paths, cp_idx)
    if use_dpd:
        T, N = pd_mat.shape
        dpd_mat = np.zeros_like(pd_mat)
        for j in range(N):
            _, dPD = condpd_to_survival_and_dpd(pd_mat[:, j])
            dpd_mat[:, j] = dPD
        y_mat = dpd_mat
    else:
        y_mat = pd_mat

    T = exp_mat.shape[0]
    uplift = np.full(T, np.nan)
    for t in range(T):
        x = exp_mat[t, :]
        y = y_mat[t, :]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() < 5:
            continue
        thresh = np.quantile(y[mask], q)
        top = mask & (y >= thresh)
        if top.sum() == 0 or np.nanmean(x[mask]) == 0:
            continue
        uplift[t] = np.nanmean(x[top]) / np.nanmean(x[mask])
    return uplift

def wwr_beta(pd_paths, exposures_model, cp_idx):
    """
    Simple regression slope of exposure on logit(PD) (per time step, then averaged).
    Returns: average beta, and a vector of betas over time.
    """
    def logit(p, eps=1e-8):
        p = np.clip(p, eps, 1-eps)
        return np.log(p/(1-p))

    exp_mat = stack_paths_exp(exposures_model, cp_idx)   # (T, N)
    pd_mat  = stack_paths_pd(pd_paths, cp_idx)           # (T, N)
    X = logit(pd_mat)
    T, N = X.shape

    betas = np.full(T, np.nan)
    for t in range(T):
        x = X[t, :]
        y = exp_mat[t, :]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() > 1 and np.std(x[mask]) > 0:
            x1 = x[mask]; y1 = y[mask]
            beta = np.dot(x1 - x1.mean(), y1 - y1.mean()) / np.dot(x1 - x1.mean(), x1 - x1.mean())
            betas[t] = beta
    return np.nanmean(betas), betas

# ===============================================================
# 3) DRIVER: RUN EVERYTHING & PRODUCE TABLES / PLOTS
# ===============================================================
def run_all_metrics(pd_paths, pd_paths_const,
                    exposures_models, model_names,
                    discount=None):
    n_cp = pd_paths[0].shape[1]
    T = pd_paths[0].shape[0]
    if discount is None:
        discount = np.ones(T)

    # --- 1) CVA WWR uplift
    uplift_mean, uplift_lo, uplift_hi = wwr_uplift_tables(
        pd_paths, pd_paths_const, exposures_models, model_names, discount
    )
    print("\nCVA WWR uplift (mean; 95% bootstrap CI in brackets):")
    for cp in uplift_mean.index:
        row = []
        for m in model_names:
            mean = uplift_mean.loc[cp, m]
            lo   = uplift_lo.loc[cp, m]
            hi   = uplift_hi.loc[cp, m]
            row.append(f"{mean:+.2%} [{lo:+.2%}, {hi:+.2%}]")
        print(cp, " | ", " | ".join(f"{m}: {v}" for m, v in zip(model_names, row)))

    # --- 2) Covariance decomposition (store totals + shares)
    cov_shares = pd.DataFrame(index=[f"CP{i+1}" for i in range(n_cp)],
                              columns=model_names, dtype=float)
    cov_totals = cov_shares.copy()
    for cp_idx in range(n_cp):
        for name, exposures in zip(model_names, exposures_models):
            br = cva_covariance_breakdown(pd_paths, exposures, cp_idx, discount)
            cov_shares.loc[f"CP{cp_idx+1}", name] = br["cov"].sum() / br["total"].sum()
            cov_totals.loc[f"CP{cp_idx+1}", name] = br["cov"].sum()

    print("\nShare of CVA coming from Cov(EXP, dPD):")
    print((cov_shares * 100).round(1).astype(str) + "%")

    # --- 3) Time-slice correlations & 4) Pathwise correlations
    pathwise_mean = pd.DataFrame(index=[f"CP{i+1}" for i in range(n_cp)],
                                 columns=model_names, dtype=float)
    pathwise_med  = pathwise_mean.copy()
    for cp_idx in range(n_cp):
        for name, exposures in zip(model_names, exposures_models):
            # Time-slice (store in a dict for optional plotting)
            rho_t = time_slice_corrs(pd_paths, exposures, cp_idx,
                                     method=METHOD_TIME_SLICE, use_dpd=False)

            # Pathwise
            r_vec, n_eff = per_sim_path_corrs(pd_paths, exposures, cp_idx,
                                              method=METHOD_PATHWISE,
                                              trim_trailing_zeros=TRIM_TRAILING_ZEROS)
            pathwise_mean.loc[f"CP{cp_idx+1}", name] = fisher_pool(r_vec, n_eff)
            pathwise_med.loc[f"CP{cp_idx+1}", name]  = np.nanmedian(r_vec)

            # Example plot (comment out if you don't want many figures)
            # plt.figure(figsize=(7,3))
            # plt.plot(rho_t)
            # plt.title(f"Time-slice {METHOD_TIME_SLICE} ρ(t) - {name} - CP{cp_idx+1}")
            # plt.xlabel("t")
            # plt.ylabel("ρ")
            # plt.grid(True, ls='--', alpha=0.4)
            # plt.tight_layout()

    print("\nPer-simulation (pathwise) Fisher-pooled correlations:")
    print(pathwise_mean.round(3))
    print("\nMedian of per-simulation correlations:")
    print(pathwise_med.round(3))

    # --- 5) Tail WWR uplift curves (example for CP1 & first model)
    cp_idx_example = 0
    model_example = 0
    uplift_curve = tail_wwr_uplift(pd_paths, exposures_models[model_example],
                                   cp_idx_example, q=TAIL_Q, use_dpd=False)
    plt.figure(figsize=(7,3))
    plt.plot(uplift_curve)
    plt.axhline(1.0, color='k', lw=1, ls='--')
    plt.title(f"Tail uplift (q={TAIL_Q:.0%}) – {model_names[model_example]} – CP{cp_idx_example+1}")
    plt.xlabel("t")
    plt.ylabel("E[EXP|tail PD] / E[EXP]")
    plt.grid(True, ls='--', alpha=0.4)
    plt.tight_layout()

    # --- 6) Optional: WWR beta
    betas_mean = pd.DataFrame(index=[f"CP{i+1}" for i in range(n_cp)],
                              columns=model_names, dtype=float)
    for cp_idx in range(n_cp):
        for name, exposures in zip(model_names, exposures_models):
            beta_mean, _ = wwr_beta(pd_paths, exposures, cp_idx)
            betas_mean.loc[f"CP{cp_idx+1}", name] = beta_mean
    print("\nAverage WWR beta (slope of EXP on logit(PD) across time):")
    print(betas_mean.round(4))

    plt.show()

    # Return everything in case you want to post-process
    return {
        "uplift_mean": uplift_mean,
        "uplift_lo": uplift_lo,
        "uplift_hi": uplift_hi,
        "cov_share": cov_shares,
        "cov_total": cov_totals,
        "pathwise_mean": pathwise_mean,
        "pathwise_median": pathwise_med,
        "betas_mean": betas_mean,
        # You can add more here if you store time-slice rhos or tail curves in dicts
    }




# Define constants
R = 0.4
NUM_SIMS = 45
NUM_STEPS = 52*5*7+2  # +2 for two leap years the coming 7 years


# Create price indexes corresponding to quarterly pricing
PRICE_POINTS = []
i = 1
number_workdays_in_quarter = (365*4+1)/7*5/4/4  # +1 for leap day every four years
while True:
    t = round(number_workdays_in_quarter*i)
    if t >= NUM_STEPS:
        break
    PRICE_POINTS.append(t)
    i = i + 1
PRICE_POINTS = np.array(PRICE_POINTS)

discount = discount_rates(PRICE_POINTS)

# ===============================================================
# 4) EXAMPLE CALL
# ===============================================================
# path = "..."  # you already have this
# (you already loaded pd_paths, exposures_en/xgb/nn/gpr)
# Now also load or construct pd_paths_const (constant-PD run):
# with open(os.path.join(path, "pd_const_YYYY-MM-DD.pkl"), "rb") as fh:
#     pd_paths_const = pickle.load(fh)

# If you DON'T have pd_paths_const, build it from a scalar per CP:
# const_pd_per_cp = [0.002, 0.003, 0.0015]  # example quarterly PDs
# pd_paths_const = make_const_pd_paths_from_scalar(pd_paths, const_pd_per_cp)

# Discount factors (example: no discounting)
# T = pd_paths[0].shape[0]
# discount = np.ones(T)

# Run:
# res = run_all_metrics(
#     pd_paths=pd_paths,
#     pd_paths_const=pd_paths_const,
#     exposures_models=[exposures_en, exposures_xgb, exposures_nn, exposures_gpr],
#     model_names=["EN", "XGB", "NN", "GPR"],
#     discount=discount
# )
```

```python
import numpy as np
import pandas as pd
from numpy.random import default_rng
from itertools import combinations
import warnings

# ============================================================
# User knobs
# ============================================================
LGD = 0.6            # you said RR=0.4
N_BOOT = 2000        # bootstrap reps
ALPHA_CI = 0.05      # 95% CI
N_PERM = 1000        # permutations for independence null
TAIL_Q = 0.975       # ES at 97.5% tail for ΔCVA
METHOD_TIME_SLICE = "spearman"  # or "pearson"
TRIM_TRAILING_ZEROS = True
RNG = default_rng(42)

# ============================================================
# -------------- Core Helpers (no SciPy needed) --------------
# ============================================================
def condpd_to_survival_and_dpd(q):
    """From a vector of *conditional* PDs q_t, compute survival S_t and unconditional default increments dPD_t."""
    q = np.clip(np.asarray(q, float), 0.0, 1.0)
    T = len(q)
    S = np.empty(T)
    dPD = np.empty(T)
    S_prev = 1.0
    for t in range(T):
        dPD[t] = S_prev * q[t]
        S[t] = S_prev * (1.0 - q[t])
        S_prev = S[t]
    return S, dPD

def make_constant_pd_paths(const_pd_per_cp, T, n_sims):
    """Build a list of n_sims DataFrames (T × n_cp) with constant conditional PD per step & counterparty."""
    n_cp = len(const_pd_per_cp)
    cols = [f"CP{i+1}" for i in range(n_cp)]
    paths = []
    for _ in range(n_sims):
        paths.append(pd.DataFrame(np.tile(const_pd_per_cp, (T, 1)), columns=cols))
    return paths

def bootstrap_ci_mean(x, n_boot=N_BOOT, alpha=ALPHA_CI, rng=RNG):
    """Percentile bootstrap CI for the mean."""
    x = np.asarray(x, float)
    n = len(x)
    boots = np.empty(n_boot)
    for b in range(n_boot):
        idx = rng.integers(0, n, n)
        boots[b] = np.mean(x[idx])
    lo, hi = np.quantile(boots, [alpha / 2, 1 - alpha / 2])
    return lo, hi

def stack_paths_pd(pd_paths, cp_idx):
    return np.stack([df.iloc[:, cp_idx].to_numpy() for df in pd_paths], axis=1)  # (T, N)

def stack_paths_exp(exposures_model, cp_idx):
    return np.stack([exp[:, cp_idx] for exp in exposures_model], axis=1)  # (T, N)

def time_slice_corrs(pd_paths, exposures_model, cp_idx, method="spearman", use_dpd=False):
    exp_mat = stack_paths_exp(exposures_model, cp_idx)
    pd_mat  = stack_paths_pd(pd_paths, cp_idx)
    if use_dpd:
        T, N = pd_mat.shape
        dpd_mat = np.zeros_like(pd_mat)
        for j in range(N):
            _, dPD = condpd_to_survival_and_dpd(pd_mat[:, j])
            dpd_mat[:, j] = dPD
        y_mat = dpd_mat
    else:
        y_mat = pd_mat

    T = exp_mat.shape[0]
    out = np.full(T, np.nan)
    for t in range(T):
        x = exp_mat[t, :]
        y = y_mat[t, :]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() > 1 and np.std(x[mask]) > 0 and np.std(y[mask]) > 0:
            if method == "pearson":
                out[t] = np.corrcoef(x[mask], y[mask])[0, 1]
            else:
                out[t] = pd.Series(x[mask]).corr(pd.Series(y[mask]), method="spearman")
    return out

def tail_uplift(pd_paths, exposures_model, cp_idx, q=0.9, use_dpd=False):
    exp_mat = stack_paths_exp(exposures_model, cp_idx)
    pd_mat  = stack_paths_pd(pd_paths, cp_idx)
    if use_dpd:
        T, N = pd_mat.shape
        dpd_mat = np.zeros_like(pd_mat)
        for j in range(N):
            _, dPD = condpd_to_survival_and_dpd(pd_mat[:, j])
            dpd_mat[:, j] = dPD
        y_mat = dpd_mat
    else:
        y_mat = pd_mat

    T = exp_mat.shape[0]
    uplift = np.full(T, np.nan)
    for t in range(T):
        x = exp_mat[t, :]
        y = y_mat[t, :]
        mask = ~(np.isnan(x) | np.isnan(y))
        if mask.sum() < 5:
            continue
        thresh = np.quantile(y[mask], q)
        top = mask & (y >= thresh)
        if top.sum() == 0 or np.nanmean(x[mask]) == 0:
            continue
        uplift[t] = np.nanmean(x[top]) / np.nanmean(x[mask])
    return uplift

def gini_from_weights(w):
    w = np.asarray(w, float)
    if w.sum() == 0:
        return np.nan
    w = np.sort(w)
    n = len(w)
    cum = np.cumsum(w)
    gini = (n + 1 - 2 * np.sum(cum / cum[-1])) / n
    return gini

# ============================================================
# ---------------- CVA engines / decomposition ---------------
# ============================================================
def cva_per_sim(exposures_model, pd_paths_like, cp_idx, discount, lgd=LGD):
    """Vector of CVA (length N_sim) using the *paired* (exposure_i, PD_i) inputs."""
    n_sim = len(exposures_model)
    out = np.zeros(n_sim)
    for i, (exp, df) in enumerate(zip(exposures_model, pd_paths_like)):
        q = df.iloc[:, cp_idx].to_numpy()
        _, dPD = condpd_to_survival_and_dpd(q)
        out[i] = lgd * np.sum(discount * exp[:, cp_idx] * dPD)
    return out

def cva_dyn_indep_vector(exposures_model, pd_paths_dyn, cp_idx, discount, lgd=LGD):
    """
    'Independence' CVA per simulation i:
        CVA_i = LGD * sum_t D_t * EXP_i(t) * E[dPD_dyn(t)]
    This keeps exposure randomness but *removes* PD–exposure dependence.
    """
    n_sim = len(exposures_model)
    # Mean dPD across sims
    q_mat = stack_paths_pd(pd_paths_dyn, cp_idx)    # (T, N)
    T, N = q_mat.shape
    dPD_mat = np.zeros_like(q_mat)
    for j in range(N):
        _, dPD = condpd_to_survival_and_dpd(q_mat[:, j])
        dPD_mat[:, j] = dPD
    mean_dPD = dPD_mat.mean(axis=1)                 # (T,)

    out = np.zeros(n_sim)
    for i, exp in enumerate(exposures_model):
        out[i] = lgd * np.sum(discount * exp[:, cp_idx] * mean_dPD)
    return out, mean_dPD

def cva_covariance_breakdown(pd_paths_dyn, exposures_model, cp_idx, discount, lgd=LGD):
    """
    Returns a DataFrame with columns: base, cov, total, share_cov per time step.
    base_t = LGD * D_t * E[EXP_t] * E[dPD_t]
    cov_t  = LGD * D_t * Cov(EXP_t, dPD_t)
    """
    exp_mat = stack_paths_exp(exposures_model, cp_idx)  # (T, N)
    q_mat   = stack_paths_pd(pd_paths_dyn, cp_idx)      # (T, N)
    T, N = q_mat.shape

    dPD_mat = np.zeros_like(q_mat)
    for j in range(N):
        _, dPD = condpd_to_survival_and_dpd(q_mat[:, j])
        dPD_mat[:, j] = dPD

    E_exp = np.nanmean(exp_mat, axis=1)
    E_dpd = np.nanmean(dPD_mat, axis=1)
    E_prod = np.nanmean(exp_mat * dPD_mat, axis=1)
    cov = E_prod - E_exp * E_dpd

    base = lgd * discount * E_exp * E_dpd
    cov_term = lgd * discount * cov
    total = base + cov_term
    share_cov = np.divide(cov_term, total, out=np.zeros_like(total), where=total != 0)

    return pd.DataFrame({
        "base": base,
        "cov": cov_term,
        "total": total,
        "share_cov": share_cov
    })

def permutation_null_test(pd_paths_dyn, exposures_model, cp_idx, discount, lgd=LGD,
                          n_perm=N_PERM, rng=RNG):
    """
    Builds the null distribution of *mean* CVA by randomly permuting PD paths across simulations,
    thereby killing dependence. Returns (obs_mean, null_means, pval, zscore).
    """
    cva_obs = cva_per_sim(exposures_model, pd_paths_dyn, cp_idx, discount, lgd).mean()

    exp_mat = stack_paths_exp(exposures_model, cp_idx)       # (T, N)
    pd_list = [df.iloc[:, cp_idx].to_numpy() for df in pd_paths_dyn]
    T, N = exp_mat.shape

    null_means = np.zeros(n_perm)
    for b in range(n_perm):
        perm = rng.permutation(N)
        tot = 0.0
        for j in range(N):
            q = pd_list[perm[j]]
            _, dPD = condpd_to_survival_and_dpd(q)
            tot += lgd * np.sum(discount * exp_mat[:, j] * dPD)
        null_means[b] = tot / N

    pval = (np.sum(null_means >= cva_obs) + 1) / (n_perm + 1)
    z = (cva_obs - null_means.mean()) / null_means.std(ddof=1)
    return cva_obs, null_means, pval, z

def frechet_bounds_cva(pd_paths_dyn, exposures_model, cp_idx, discount, lgd=LGD):
    """
    Theoretical upper/lower bounds across time by pairing (per time step) sorted exposures
    with sorted dPDs (same order for max, reverse for min).
    """
    exp_mat = stack_paths_exp(exposures_model, cp_idx)  # (T, N)
    q_mat   = stack_paths_pd(pd_paths_dyn, cp_idx)      # (T, N)
    T, N = q_mat.shape

    dPD_mat = np.zeros_like(q_mat)
    for j in range(N):
        _, dPD = condpd_to_survival_and_dpd(q_mat[:, j])
        dPD_mat[:, j] = dPD

    cva_upper = 0.0
    cva_lower = 0.0
    cva_indep = 0.0
    for t in range(T):
        e = np.sort(exp_mat[t, :])
        d = np.sort(dPD_mat[t, :])
        cva_upper += lgd * discount[t] * np.dot(e, d)          # comonotonic
        cva_lower += lgd * discount[t] * np.dot(e, d[::-1])    # countermonotonic
        cva_indep += lgd * discount[t] * (exp_mat[t, :].mean() * dPD_mat[t, :].mean())

    return cva_lower, cva_indep, cva_upper

def es_allocation_delta_cva(pd_paths_dyn, exposures_model, pd_paths_static,
                            cp_idx, discount, lgd=LGD, tail_q=TAIL_Q):
    """
    ES of ΔCVA (dyn joint - static) and its time allocation using covariance weights
    computed on the tail simulations only.
    """
    cva_joint = cva_per_sim(exposures_model, pd_paths_dyn, cp_idx, discount, lgd)
    cva_stat  = cva_per_sim(exposures_model, pd_paths_static, cp_idx, discount, lgd)
    prem = cva_joint - cva_stat

    thresh = np.quantile(prem, tail_q)
    tail_mask = prem >= thresh
    ES = prem[tail_mask].mean()

    # Covariance weights on tail only
    exp_mat = stack_paths_exp(exposures_model, cp_idx)[:, tail_mask]
    q_mat   = stack_paths_pd(pd_paths_dyn, cp_idx)[:, tail_mask]
    T, N_tail = q_mat.shape
    dPD_mat = np.zeros_like(q_mat)
    for j in range(N_tail):
        _, dPD = condpd_to_survival_and_dpd(q_mat[:, j])
        dPD_mat[:, j] = dPD

    E_exp  = np.nanmean(exp_mat, axis=1)
    E_dpd  = np.nanmean(dPD_mat, axis=1)
    E_prod = np.nanmean(exp_mat * dPD_mat, axis=1)
    cov    = E_prod - E_exp * E_dpd
    cov_term = lgd * discount * cov
    weights = np.maximum(cov_term, 0)
    if weights.sum() == 0:
        weights[:] = 1.0
    weights /= weights.sum()

    alloc = ES * weights
    return ES, alloc, weights

# ============================================================
# --------------- Main orchestrator (report) -----------------
# ============================================================
def run_full_wwr_analysis(
        cva_models,                    # list[pd.DataFrame] OR dict[str, pd.DataFrame]
        exposures_models,              # list like [exposures_en, exposures_xgb, exposures_nn, exposures_gpr]
        pd_paths_dyn,                  # list of DataFrames (n_sims)
        const_pd_per_cp,               # list/array length n_cp of CONSTANT conditional PD per step
        discount,                      # np.array length T
        model_names=None,
        lgd=LGD,
        n_boot=N_BOOT,
        n_perm=N_PERM,
        alpha_ci=ALPHA_CI,
        tail_q=TAIL_Q,
        method_time_slice=METHOD_TIME_SLICE
    ):
    """
    Returns a dict of DataFrames and dicts containing every metric / decomposition we discussed.
    """

    # ---------------- harmonise model names & CVA DFs ----------------
    if isinstance(cva_models, dict):
        model_names = list(cva_models.keys()) if model_names is None else model_names
        cva_dict = cva_models
    else:
        if model_names is None:
            model_names = ["EN", "XGB", "NN", "GPR"][:len(cva_models)]
        cva_dict = {name: df.copy() for name, df in zip(model_names, cva_models)}

    n_models = len(model_names)
    n_sims   = next(iter(cva_dict.values())).shape[0]
    n_cp     = next(iter(cva_dict.values())).shape[1]
    cp_names = list(next(iter(cva_dict.values())).columns)

    # ---------------- build static PD paths ----------------
    T = pd_paths_dyn[0].shape[0]
    assert len(discount) == T, "discount must have length T"
    assert len(const_pd_per_cp) == n_cp, "Provide one constant conditional PD per counterparty"
    pd_paths_static = make_constant_pd_paths(const_pd_per_cp, T, n_sims)

    # ---------------- containers for results ----------------
    # Final summary table
    cols = [
        "CVA_static_mean", "CVA_dyn_indep_mean", "CVA_dyn_joint_mean",
        "Δ_total", "Δ_level_timing", "Δ_WWR",
        "%_level_timing_of_total", "%_WWR_of_total"
    ]
    summary = pd.DataFrame(
        index=pd.MultiIndex.from_product([cp_names, model_names], names=["counterparty", "model"]),
        columns=cols, dtype=float
    )

    # CIs
    ci_cols = [c + "_lo" for c in cols[:3]] + [c + "_hi" for c in cols[:3]]
    ci = pd.DataFrame(index=summary.index, columns=ci_cols, dtype=float)

    # permutation test p-values/z
    perm_df = pd.DataFrame(index=summary.index, columns=["p_value", "z_score"], dtype=float)

    # time profiles, correlations, tail uplift, ES allocations, bounds
    time_profiles = {}        # (cp, model) -> DataFrame with base, cov, total, share_cov
    rho_time = {}             # (cp, model) -> np.array
    tail_uplifts = {}         # (cp, model) -> np.array
    es_allocations = {}       # (cp, model) -> dict(ES, alloc, weights)
    bounds_tbl = {}           # (cp, model) -> dict(lower, indep, upper)

    # ---------------- run per CP × model ----------------
    for cp_idx, cp_name in enumerate(cp_names):
        for m_idx, model in enumerate(model_names):
            exposures = exposures_models[m_idx]

            # 1) CVAs per sim
            cva_static_vec = cva_per_sim(exposures, pd_paths_static, cp_idx, discount, lgd)
            cva_dyn_joint_vec = cva_per_sim(exposures, pd_paths_dyn, cp_idx, discount, lgd)
            cva_dyn_indep_vec, mean_dPD = cva_dyn_indep_vector(exposures, pd_paths_dyn, cp_idx, discount, lgd)

            # 2) Means
            mu_static = cva_static_vec.mean()
            mu_joint  = cva_dyn_joint_vec.mean()
            mu_indep  = cva_dyn_indep_vec.mean()

            # 3) Effects
            delta_total  = mu_joint - mu_static
            delta_level  = mu_indep - mu_static
            delta_wwr    = mu_joint - mu_indep
            pct_level    = delta_level / delta_total if delta_total != 0 else np.nan
            pct_wwr      = delta_wwr   / delta_total if delta_total != 0 else np.nan

            # 4) CIs (bootstrap on per-sim vectors)
            lo_stat, hi_stat = bootstrap_ci_mean(cva_static_vec, n_boot=n_boot, alpha=alpha_ci, rng=RNG)
            lo_ind,  hi_ind  = bootstrap_ci_mean(cva_dyn_indep_vec, n_boot=n_boot, alpha=alpha_ci, rng=RNG)
            lo_joint,hi_joint= bootstrap_ci_mean(cva_dyn_joint_vec, n_boot=n_boot, alpha=alpha_ci, rng=RNG)

            # 5) permutation test (null: no dependence)
            cva_obs, null_means, pval, z = permutation_null_test(
                pd_paths_dyn, exposures, cp_idx, discount, lgd, n_perm=n_perm, rng=RNG
            )

            # 6) time profile (covariance decomposition)
            prof = cva_covariance_breakdown(pd_paths_dyn, exposures, cp_idx, discount, lgd)

            # 7) time-slice correlations (across sims)
            rho = time_slice_corrs(pd_paths_dyn, exposures, cp_idx, method=method_time_slice, use_dpd=False)

            # 8) tail uplift curve (PD tail)
            tu = tail_uplift(pd_paths_dyn, exposures, cp_idx, q=0.9, use_dpd=False)

            # 9) ES allocation of ΔCVA
            ES, alloc, weights = es_allocation_delta_cva(pd_paths_dyn, exposures, pd_paths_static,
                                                         cp_idx, discount, lgd, tail_q=tail_q)

            # 10) Fréchet bounds
            cva_low, cva_ind, cva_up = frechet_bounds_cva(pd_paths_dyn, exposures, cp_idx, discount, lgd)

            # -------- store
            idx = (cp_name, model)
            summary.loc[idx] = [
                mu_static, mu_indep, mu_joint,
                delta_total, delta_level, delta_wwr,
                pct_level * 100, pct_wwr * 100
            ]
            ci.loc[idx] = [lo_stat, lo_ind, lo_joint, hi_stat, hi_ind, hi_joint]
            perm_df.loc[idx] = [pval, z]
            time_profiles[idx] = prof
            rho_time[idx] = rho
            tail_uplifts[idx] = tu
            es_allocations[idx] = {"ES": ES, "alloc": alloc, "weights": weights}
            bounds_tbl[idx] = {"lower": cva_low, "indep_proxy": cva_ind, "upper": cva_up}

    # Nicely order and round
    summary = summary.astype(float)
    ci = ci.astype(float)
    perm_df = perm_df.astype(float)

    return {
        "summary": summary,
        "ci": ci,
        "perm_test": perm_df,
        "time_profiles": time_profiles,
        "rho_time": rho_time,
        "tail_uplifts": tail_uplifts,
        "es_allocations": es_allocations,
        "bounds": bounds_tbl
    }
```
